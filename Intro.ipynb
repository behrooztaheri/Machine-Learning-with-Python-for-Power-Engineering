# While

```python
def Call_int():
    while True:
        try:
            NUM = int(input("Please enter a num: "))
            return NUM
        except:
            print("Please enter a NUM")

def Call_float():
    while True:
        try:
            NUM = float(input("Please enter a num: "))
            return NUM
        except:
            print("Please enter a NUM")
```

```python
if Call_int() % 2 == 0:
    print("Zoj")
else:
    print("Fard")
```
