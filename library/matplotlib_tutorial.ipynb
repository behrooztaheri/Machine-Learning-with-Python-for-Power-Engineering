## 1. Why Matplotlib?

**Matplotlib** is the most widely used plotting library in Python. Libraries such as **Seaborn, Pandas plotting, and scikit-learn's display utilities** are built on top of it.

In AI and data science you use visualization to:
- **Explore data** (distributions, outliers, correlations)
- **Monitor training** (loss and accuracy curves)
- **Evaluate models** (confusion matrix, ROC curve)
- **Understand models** (decision boundaries, feature maps, embeddings)
- **Communicate results** (papers, reports, presentations)

### Anatomy of a figure
- **Figure:** the whole canvas (the window or image).
- **Axes:** a single plot area inside the figure (it holds the x-axis, y-axis, title, and data). One figure can contain many axes.
- **Axis:** the x or y number line.
- **Artist:** everything you see (lines, text, ticks, legends, ...).

### Two interfaces
| Interface | Style | When to use |
|---|---|---|
| `pyplot` (state-based) | `plt.plot(...)`, `plt.title(...)` | Quick experiments |
| Object-oriented | `fig, ax = plt.subplots()`, `ax.plot(...)` | Real projects, multiple plots (**recommended**) |

## 3. Your First Plot

`plt.plot(x, y)` draws a line. `plt.show()` displays the figure.

````python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("My First Plot: sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
````

Plot several lines on the same axes, add a legend and a grid.

````python
x = np.linspace(0, 2 * np.pi, 200)

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.title("Sine and Cosine")
plt.xlabel("Angle (radians)")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()
````

## 4. Customizing Lines, Markers, and Colors

### Format string shortcut: `"[marker][line][color]"`
Examples: `"ro--"` = red circles with a dashed line, `"g^:"` = green triangles with a dotted line.

### Common keyword arguments
| Argument | Meaning | Examples |
|---|---|---|
| `color` / `c` | Line color | `"red"`, `"#1f77b4"`, `(0.2, 0.4, 0.6)` |
| `linestyle` / `ls` | Line style | `"-"`, `"--"`, `"-."`, `":"` |
| `linewidth` / `lw` | Thickness | `2` |
| `marker` | Point marker | `"o"`, `"s"`, `"^"`, `"*"`, `"x"` |
| `markersize` / `ms` | Marker size | `8` |
| `alpha` | Transparency (0 to 1) | `0.5` |
| `label` | Legend text | `"Model A"` |

````python
x = np.arange(1, 8)

plt.figure(figsize=(8, 4))
plt.plot(x, x,       "ro--", label="linear")
plt.plot(x, x ** 1.5, color="green", linestyle="-.", marker="s", label="power 1.5")
plt.plot(x, x ** 2,  color="#1f77b4", linewidth=3, marker="^", markersize=9, alpha=0.7, label="quadratic")
plt.legend(loc="upper left")
plt.title("Line Customization")
plt.grid(alpha=0.3)
plt.show()
````

## 5. Figure and Axes: The Object-Oriented Interface

This is the recommended approach. `plt.subplots()` returns a **figure** and an **axes** object. You then call methods on `ax` such as `ax.plot`, `ax.set_title`, `ax.set_xlabel`.

````python
x = np.linspace(0, 10, 100)

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, np.exp(-x / 3) * np.cos(2 * x), color="purple", lw=2, label="damped cosine")
ax.set_title("Object-Oriented Interface")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Amplitude")
ax.legend()
ax.grid(alpha=0.3)

plt.show()
````

> **Naming note:** with `pyplot` you call `plt.title()`, but with Axes objects the method is `ax.set_title()`. Same for `set_xlabel`, `set_ylabel`, `set_xlim`, `set_ylim`.

## 6. Multiple Plots with Subplots

`plt.subplots(nrows, ncols)` creates a grid of axes (returned as a NumPy array).

````python
x = np.linspace(0, 2 * np.pi, 200)

fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0, 0].plot(x, np.sin(x), color="tab:blue")
axes[0, 0].set_title("sin(x)")

axes[0, 1].plot(x, np.cos(x), color="tab:orange")
axes[0, 1].set_title("cos(x)")

axes[1, 0].plot(x, np.tan(x), color="tab:green")
axes[1, 0].set_ylim(-5, 5)
axes[1, 0].set_title("tan(x) (clipped)")

axes[1, 1].plot(x, np.sin(x) ** 2, color="tab:red")
axes[1, 1].set_title("sin^2(x)")

for ax in axes.flat:
    ax.grid(alpha=0.3)

fig.suptitle("A 2x2 Grid of Subplots", fontsize=14)
fig.tight_layout()
plt.show()
````

### Shared axes and unequal layouts
Use `sharex=True` / `sharey=True` to share axes. For uneven layouts use `plt.subplot_mosaic`.

````python
x = np.linspace(0, 10, 200)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 3.5), sharey=True)
ax1.plot(x, np.sin(x));       ax1.set_title("Left")
ax2.plot(x, np.sin(2 * x));   ax2.set_title("Right (shares y-axis)")
fig.tight_layout()
plt.show()

# Mosaic layout: one wide plot on top, two below
fig, axd = plt.subplot_mosaic([["top", "top"],
                               ["bl",  "br"]], figsize=(8, 6))
axd["top"].plot(x, np.sin(x));  axd["top"].set_title("Top (wide)")
axd["bl"].plot(x, np.cos(x));   axd["bl"].set_title("Bottom left")
axd["br"].plot(x, x ** 0.5);    axd["br"].set_title("Bottom right")
fig.tight_layout()
plt.show()
````

## 7. Scatter Plots

Scatter plots show the relationship between two variables. They are essential for visualizing datasets, clusters, and embeddings.

Useful arguments: `s` (size), `c` (color or values), `cmap` (colormap), `alpha`, `edgecolor`.

````python
rng = np.random.default_rng(0)
x = rng.normal(0, 1, 200)
y = 2 * x + rng.normal(0, 1, 200)

fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(x, y, alpha=0.6, edgecolor="k")
ax.set_title("Basic Scatter Plot")
ax.set_xlabel("Feature x")
ax.set_ylabel("Target y")
plt.show()
````

### Encoding extra information with color and size

````python
rng = np.random.default_rng(1)
n = 150
x = rng.random(n)
y = rng.random(n)
colors = x + y                    # a numeric value mapped to color
sizes = 20 + 300 * rng.random(n)  # marker area

fig, ax = plt.subplots(figsize=(7, 5))
sc = ax.scatter(x, y, c=colors, s=sizes, cmap="viridis", alpha=0.7, edgecolor="white")
fig.colorbar(sc, ax=ax, label="x + y")
ax.set_title("Color and Size Encoding")
plt.show()
````

### Plotting classes with different colors (typical for classification)

````python
rng = np.random.default_rng(2)
class0 = rng.normal([-2, -2], 1.0, size=(100, 2))
class1 = rng.normal([ 2,  2], 1.0, size=(100, 2))
class2 = rng.normal([-2,  3], 1.0, size=(100, 2))

fig, ax = plt.subplots(figsize=(6, 5))
for data, name, color in [(class0, "Class 0", "tab:blue"),
                          (class1, "Class 1", "tab:orange"),
                          (class2, "Class 2", "tab:green")]:
    ax.scatter(data[:, 0], data[:, 1], label=name, color=color, alpha=0.7, edgecolor="k", linewidth=0.3)

ax.legend()
ax.set_title("Three-Class Dataset")
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
plt.show()
````

## 8. Bar Charts

Use `ax.bar` (vertical) or `ax.barh` (horizontal) to compare categories, for example model accuracies.

````python
models = ["Logistic\nRegression", "SVM", "Random\nForest", "Neural\nNet"]
accuracy = [0.82, 0.86, 0.91, 0.94]

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(models, accuracy, color=["#8ecae6", "#219ebc", "#126782", "#023047"])
ax.set_ylim(0.7, 1.0)
ax.set_ylabel("Accuracy")
ax.set_title("Model Comparison")

# Write the value above each bar
for bar, value in zip(bars, accuracy):
    ax.text(bar.get_x() + bar.get_width() / 2, value + 0.005, f"{value:.2f}",
            ha="center", va="bottom", fontweight="bold")
plt.show()
````

### Grouped and stacked bars

````python
labels = ["Model A", "Model B", "Model C"]
precision = np.array([0.80, 0.85, 0.90])
recall    = np.array([0.75, 0.88, 0.82])

x = np.arange(len(labels))
width = 0.35

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

# Grouped
ax1.bar(x - width / 2, precision, width, label="Precision")
ax1.bar(x + width / 2, recall,    width, label="Recall")
ax1.set_xticks(x)
ax1.set_xticklabels(labels)
ax1.set_title("Grouped Bars")
ax1.legend()

# Stacked
ax2.bar(labels, precision, label="Precision")
ax2.bar(labels, recall, bottom=precision, label="Recall")
ax2.set_title("Stacked Bars")
ax2.legend()

fig.tight_layout()
plt.show()
````

### Horizontal bar chart (great for feature importance)

````python
features = ["Age", "Income", "Education", "Hours/week", "Savings"]
importance = np.array([0.12, 0.35, 0.20, 0.08, 0.25])

order = np.argsort(importance)       # sort ascending so the biggest is on top

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.barh(np.array(features)[order], importance[order], color="teal")
ax.set_xlabel("Importance")
ax.set_title("Feature Importance")
plt.show()
````

## 9. Histograms

A histogram shows the **distribution** of a numeric variable. Key arguments: `bins`, `density`, `alpha`, `histtype`.

````python
rng = np.random.default_rng(3)
data = rng.normal(loc=50, scale=10, size=2000)

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(data, bins=30, color="skyblue", edgecolor="black")
ax.axvline(data.mean(), color="red", linestyle="--", label=f"Mean = {data.mean():.1f}")
ax.set_title("Histogram of Exam Scores")
ax.set_xlabel("Score")
ax.set_ylabel("Frequency")
ax.legend()
plt.show()
````

### Comparing distributions
Use `density=True` to normalize so that distributions with different sample sizes are comparable.

````python
rng = np.random.default_rng(4)
a = rng.normal(0, 1, 3000)
b = rng.normal(1.5, 1.5, 1000)

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(a, bins=40, alpha=0.6, density=True, label="Group A")
ax.hist(b, bins=40, alpha=0.6, density=True, label="Group B")
ax.set_title("Overlapping Histograms (normalized)")
ax.legend()
plt.show()
````

### 2D histogram (density of points)

````python
rng = np.random.default_rng(5)
x = rng.normal(0, 1, 20000)
y = x * 0.5 + rng.normal(0, 1, 20000)

fig, ax = plt.subplots(figsize=(6, 5))
h = ax.hist2d(x, y, bins=50, cmap="magma")
fig.colorbar(h[3], ax=ax, label="Count")
ax.set_title("2D Histogram")
plt.show()
````

## 10. Pie Charts

Pie charts show proportions of a whole. Use them sparingly (bar charts are usually easier to read).

````python
labels = ["Training", "Validation", "Test"]
sizes = [70, 15, 15]
explode = (0.05, 0, 0)

fig, ax = plt.subplots(figsize=(5, 5))
ax.pie(sizes, labels=labels, explode=explode, autopct="%1.0f%%",
       startangle=90, colors=["#06d6a0", "#ffd166", "#ef476f"],
       wedgeprops={"edgecolor": "white"})
ax.set_title("Dataset Split")
plt.show()
````

## 11. Box Plots and Violin Plots

Box plots summarize a distribution using the **median, quartiles, and outliers**. Violin plots additionally show the density shape.

````python
rng = np.random.default_rng(6)
data = [rng.normal(0, 1, 200),
        rng.normal(1, 2, 200),
        rng.normal(-1, 0.5, 200)]
names = ["A", "B", "C"]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4), sharey=True)

ax1.boxplot(data)
ax1.set_xticklabels(names)
ax1.set_title("Box Plot")

ax2.violinplot(data, showmedians=True)
ax2.set_xticks([1, 2, 3])
ax2.set_xticklabels(names)
ax2.set_title("Violin Plot")

fig.tight_layout()
plt.show()
````

## 12. Heatmaps and Images

`ax.imshow(array)` displays a 2D array as an image or heatmap. This is used for images, correlation matrices, attention maps, and confusion matrices.

### Correlation matrix heatmap

````python
rng = np.random.default_rng(7)
X = rng.normal(size=(200, 4))
X[:, 1] = X[:, 0] * 0.8 + rng.normal(0, 0.5, 200)    # make feature 1 correlated with 0
X[:, 3] = -X[:, 2] * 0.6 + rng.normal(0, 0.7, 200)   # negative correlation

corr = np.corrcoef(X, rowvar=False)
names = ["f0", "f1", "f2", "f3"]

fig, ax = plt.subplots(figsize=(5, 4.5))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
fig.colorbar(im, ax=ax, label="Correlation")

ax.set_xticks(range(4)); ax.set_xticklabels(names)
ax.set_yticks(range(4)); ax.set_yticklabels(names)

# Write values in cells
for i in range(4):
    for j in range(4):
        ax.text(j, i, f"{corr[i, j]:.2f}", ha="center", va="center", color="black")

ax.set_title("Correlation Matrix")
plt.show()
````

### Displaying images
Grayscale images are 2D arrays; color images are `(H, W, 3)` arrays. Here we build synthetic images with NumPy.

````python
# Synthetic grayscale image: a smooth gradient with a bright circle
H, W = 64, 64
yy, xx = np.mgrid[0:H, 0:W]
gradient = xx / W
circle = ((xx - 32) ** 2 + (yy - 32) ** 2) < 15 ** 2
img_gray = np.clip(gradient + circle * 0.8, 0, 1)

# Synthetic RGB image
img_rgb = np.stack([gradient, yy / H, circle.astype(float)], axis=-1)

fig, axes = plt.subplots(1, 3, figsize=(11, 3.8))

axes[0].imshow(img_gray, cmap="gray")
axes[0].set_title("Grayscale (cmap='gray')")

axes[1].imshow(img_gray, cmap="viridis")
axes[1].set_title("Same data (cmap='viridis')")

axes[2].imshow(img_rgb)
axes[2].set_title("RGB image")

for ax in axes:
    ax.axis("off")
fig.tight_layout()
plt.show()
````

## 13. Contour and 3D Plots

### Contour plots
Use `np.meshgrid` to create a grid, evaluate a function on it, and draw `contour` (lines) or `contourf` (filled). Contour plots are the standard way to visualize **loss surfaces** and **decision boundaries**.

````python
x = np.linspace(-3, 3, 200)
y = np.linspace(-3, 3, 200)
X, Y = np.meshgrid(x, y)
Z = X ** 2 + 3 * Y ** 2          # an elongated bowl, like a loss surface

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5))

cs = ax1.contour(X, Y, Z, levels=15, cmap="viridis")
ax1.clabel(cs, inline=True, fontsize=7)
ax1.set_title("contour")

cf = ax2.contourf(X, Y, Z, levels=20, cmap="viridis")
fig.colorbar(cf, ax=ax2, label="Loss")
ax2.set_title("contourf")

for ax in (ax1, ax2):
    ax.set_xlabel("w1")
    ax.set_ylabel("w2")
fig.tight_layout()
plt.show()
````

### 3D surface plot

````python
x = np.linspace(-3, 3, 60)
y = np.linspace(-3, 3, 60)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X ** 2 + Y ** 2))

fig = plt.figure(figsize=(7, 5.5))
ax = fig.add_subplot(111, projection="3d")
surf = ax.plot_surface(X, Y, Z, cmap="plasma", edgecolor="none")
fig.colorbar(surf, ax=ax, shrink=0.6)
ax.set_title("3D Surface")
ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
plt.show()
````

### 3D scatter

````python
rng = np.random.default_rng(8)
pts = rng.normal(size=(300, 3))
labels = (pts[:, 0] + pts[:, 1] > 0).astype(int)

fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(pts[:, 0], pts[:, 1], pts[:, 2], c=labels, cmap="coolwarm", alpha=0.8)
ax.set_title("3D Scatter")
plt.show()
````

## 14. Error Bars and Filled Regions

Showing **uncertainty** is vital in ML (for example, averaging results over several runs).

### Error bars

````python
x = np.arange(1, 6)
mean = np.array([0.72, 0.78, 0.83, 0.86, 0.88])
std = np.array([0.05, 0.04, 0.03, 0.03, 0.02])

fig, ax = plt.subplots(figsize=(6, 4))
ax.errorbar(x, mean, yerr=std, fmt="o-", capsize=5, color="tab:blue", label="Mean +/- std")
ax.set_xlabel("Number of layers")
ax.set_ylabel("Accuracy")
ax.set_title("Error Bars")
ax.legend()
plt.show()
````

### Shaded confidence band with `fill_between`
Simulate 10 training runs and plot the mean curve with a band of one standard deviation.

````python
rng = np.random.default_rng(9)
epochs = np.arange(1, 51)
runs = np.array([np.exp(-epochs / 15) + rng.normal(0, 0.03, len(epochs)) for _ in range(10)])

mean = runs.mean(axis=0)
std = runs.std(axis=0)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(epochs, mean, color="tab:red", label="Mean loss (10 runs)")
ax.fill_between(epochs, mean - std, mean + std, color="tab:red", alpha=0.25, label="+/- 1 std")
ax.set_xlabel("Epoch")
ax.set_ylabel("Loss")
ax.set_title("Mean and Variability Across Runs")
ax.legend()
plt.show()
````

## 15. Annotations, Scales, Ticks, and Limits

### Text, arrows, and reference lines

````python
x = np.linspace(0, 10, 200)
y = np.sin(x) * np.exp(-x / 5)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, y, lw=2)

peak_idx = np.argmax(y)
ax.annotate("Peak", xy=(x[peak_idx], y[peak_idx]),
            xytext=(x[peak_idx] + 2, y[peak_idx] + 0.1),
            arrowprops=dict(arrowstyle="->", color="red"),
            fontsize=12, color="red")

ax.axhline(0, color="gray", linestyle="--")      # horizontal reference line
ax.axvline(np.pi, color="green", linestyle=":")  # vertical reference line
ax.axvspan(6, 8, color="orange", alpha=0.2)      # highlighted vertical band
ax.text(6.2, -0.35, "Region of interest", color="darkorange")

ax.set_title("Annotations and Reference Lines")
plt.show()
````

### Logarithmic scales
Log scales are useful when values span several orders of magnitude (for example, learning rates or loss values).

````python
x = np.logspace(-4, 0, 50)
y = x ** 2

fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))

axes[0].plot(x, y);        axes[0].set_title("Linear scale")
axes[1].plot(x, y);        axes[1].set_xscale("log");  axes[1].set_title("Log x")
axes[2].plot(x, y);        axes[2].set_xscale("log");  axes[2].set_yscale("log"); axes[2].set_title("Log-log")

fig.tight_layout()
plt.show()
````

### Ticks, limits, and LaTeX-style math text

````python
x = np.linspace(-2 * np.pi, 2 * np.pi, 300)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, np.sin(x), label=r"$\sin(x)$")
ax.plot(x, np.cos(x), label=r"$\cos(x)$")

ax.set_xlim(-2 * np.pi, 2 * np.pi)
ax.set_ylim(-1.5, 1.5)
ax.set_xticks([-2 * np.pi, -np.pi, 0, np.pi, 2 * np.pi])
ax.set_xticklabels([r"$-2\pi$", r"$-\pi$", "0", r"$\pi$", r"$2\pi$"])
ax.set_title(r"Trigonometric functions: $y = \sin(x)$ and $y = \cos(x)$")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
````

### Colormaps
Choose colormaps carefully:
- **Sequential** (`viridis`, `plasma`, `Blues`): ordered data.
- **Diverging** (`coolwarm`, `RdBu`): data with a meaningful midpoint such as zero (for example, correlations).
- **Qualitative** (`tab10`, `Set2`): categories.
- Avoid `jet`; it is perceptually misleading.

````python
gradient = np.linspace(0, 1, 256).reshape(1, -1)
gradient = np.vstack([gradient] * 12)

cmaps = ["viridis", "plasma", "coolwarm", "Blues", "tab10", "gray"]
fig, axes = plt.subplots(len(cmaps), 1, figsize=(7, 4))

for ax, name in zip(axes, cmaps):
    ax.imshow(gradient, aspect="auto", cmap=name)
    ax.set_axis_off()
    ax.text(-0.01, 0.5, name, va="center", ha="right", transform=ax.transAxes, fontsize=10)

plt.show()
````

## 17. Saving Figures

Use `fig.savefig()` with a file name. The extension decides the format (`.png`, `.jpg`, `.svg`, `.pdf`).

- `dpi`: resolution (use 300 for papers).

### Project 2: Training and validation curves
Plot loss and accuracy per epoch to detect **overfitting** (training loss keeps falling while validation loss rises).

````python
rng = np.random.default_rng(10)
epochs = np.arange(1, 41)

train_loss = 1.5 * np.exp(-epochs / 8) + 0.05 + rng.normal(0, 0.01, len(epochs))
val_loss   = 1.5 * np.exp(-epochs / 8) + 0.15 + 0.004 * np.maximum(0, epochs - 18) ** 1.3 \
             + rng.normal(0, 0.02, len(epochs))

train_acc = 1 - 0.6 * np.exp(-epochs / 8) - 0.02
val_acc   = 1 - 0.6 * np.exp(-epochs / 8) - 0.08 - 0.002 * np.maximum(0, epochs - 18)

best_epoch = int(np.argmin(val_loss)) + 1

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(epochs, train_loss, label="Train loss")
ax1.plot(epochs, val_loss, label="Validation loss")
ax1.axvline(best_epoch, color="gray", linestyle="--", label=f"Best epoch = {best_epoch}")
ax1.set_xlabel("Epoch"); ax1.set_ylabel("Loss"); ax1.set_title("Loss")
ax1.legend()

ax2.plot(epochs, train_acc, label="Train accuracy")
ax2.plot(epochs, val_acc, label="Validation accuracy")
ax2.set_xlabel("Epoch"); ax2.set_ylabel("Accuracy"); ax2.set_title("Accuracy")
ax2.legend()

fig.suptitle("Training Curves (overfitting after the dashed line)", fontsize=13)
fig.tight_layout()
plt.show()
````

### Project 3: Confusion matrix

````python
def confusion_matrix(y_true, y_pred, num_classes):
    cm = np.zeros((num_classes, num_classes), dtype=int)
    np.add.at(cm, (y_true, y_pred), 1)
    return cm

rng = np.random.default_rng(11)
classes = ["cat", "dog", "bird"]
y_true = rng.integers(0, 3, size=300)
# Simulate a classifier that is right ~75% of the time
noise = rng.integers(0, 3, size=300)
y_pred = np.where(rng.random(300) < 0.75, y_true, noise)

cm = confusion_matrix(y_true, y_pred, 3)

fig, ax = plt.subplots(figsize=(5, 4.5))
im = ax.imshow(cm, cmap="Blues")
fig.colorbar(im, ax=ax)

ax.set_xticks(range(3)); ax.set_xticklabels(classes)
ax.set_yticks(range(3)); ax.set_yticklabels(classes)
ax.set_xlabel("Predicted label")
ax.set_ylabel("True label")
ax.set_title(f"Confusion Matrix (accuracy = {np.trace(cm) / cm.sum():.2%})")

threshold = cm.max() / 2
for i in range(3):
    for j in range(3):
        ax.text(j, i, cm[i, j], ha="center", va="center",
                color="white" if cm[i, j] > threshold else "black", fontsize=13)
plt.show()
````

### Project 4: ROC curve and AUC

````python
rng = np.random.default_rng(12)
n = 500
y_true = rng.integers(0, 2, size=n)
# Scores: positives tend to have higher scores
scores = np.where(y_true == 1, rng.normal(0.65, 0.2, n), rng.normal(0.4, 0.2, n))

# Compute ROC manually
thresholds = np.linspace(1.2, -0.2, 200)
tpr, fpr = [], []
P = (y_true == 1).sum()
N = (y_true == 0).sum()
for t in thresholds:
    pred = scores >= t
    tpr.append(((pred) & (y_true == 1)).sum() / P)
    fpr.append(((pred) & (y_true == 0)).sum() / N)
tpr, fpr = np.array(tpr), np.array(fpr)

auc = np.sum(np.diff(fpr) * (tpr[1:] + tpr[:-1]) / 2)   # trapezoidal rule

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(fpr, tpr, lw=2, label=f"Model (AUC = {auc:.3f})")
ax.plot([0, 1], [0, 1], "k--", label="Random guess")
ax.fill_between(fpr, tpr, alpha=0.15)
ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")
ax.set_title("ROC Curve")
ax.legend(loc="lower right")
ax.set_aspect("equal")
plt.show()
````

### Project 5: Decision boundary of a classifier
We train **logistic regression** with gradient descent (NumPy only) on 2D data and draw its decision boundary with `contourf`.

````python
rng = np.random.default_rng(13)
n = 150
X0 = rng.normal([-1.5, -1.0], 1.0, size=(n, 2))
X1 = rng.normal([ 1.5,  1.0], 1.0, size=(n, 2))
X = np.vstack([X0, X1])
y = np.array([0] * n + [1] * n)

# Logistic regression with gradient descent
Xb = np.column_stack([np.ones(len(X)), X])
w = np.zeros(3)
sigmoid = lambda z: 1 / (1 + np.exp(-z))
for _ in range(2000):
    p = sigmoid(Xb @ w)
    w -= 0.1 * Xb.T @ (p - y) / len(y)

# Evaluate the model on a dense grid
xx, yy = np.meshgrid(np.linspace(X[:, 0].min() - 1, X[:, 0].max() + 1, 300),
                     np.linspace(X[:, 1].min() - 1, X[:, 1].max() + 1, 300))
grid = np.column_stack([np.ones(xx.size), xx.ravel(), yy.ravel()])
probs = sigmoid(grid @ w).reshape(xx.shape)

fig, ax = plt.subplots(figsize=(6.5, 5))
cf = ax.contourf(xx, yy, probs, levels=20, cmap="RdBu_r", alpha=0.7)
ax.contour(xx, yy, probs, levels=[0.5], colors="k", linewidths=2)   # the boundary p = 0.5
fig.colorbar(cf, ax=ax, label="P(class = 1)")
ax.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k", linewidth=0.4)
ax.set_title("Logistic Regression Decision Boundary")
ax.set_xlabel("Feature 1"); ax.set_ylabel("Feature 2")

acc = ((sigmoid(Xb @ w) > 0.5) == y).mean()
print(f"Training accuracy: {acc:.2%}")
plt.show()
````

### Project 6: Gradient descent on a loss surface
Visualize how gradient descent moves across a contour plot, and how the learning rate changes the path.

````python
def loss(w):
    return w[0] ** 2 + 5 * w[1] ** 2

def grad(w):
    return np.array([2 * w[0], 10 * w[1]])

def run_gd(lr, steps=30, start=(4.0, 2.0)):
    w = np.array(start)
    path = [w.copy()]
    for _ in range(steps):
        w = w - lr * grad(w)
        path.append(w.copy())
    return np.array(path)

x = np.linspace(-5, 5, 200)
y = np.linspace(-3, 3, 200)
X, Y = np.meshgrid(x, y)
Z = X ** 2 + 5 * Y ** 2

fig, axes = plt.subplots(1, 3, figsize=(14, 4.2), sharey=True)
for ax, lr in zip(axes, [0.02, 0.15, 0.21]):
    ax.contour(X, Y, Z, levels=20, cmap="Greys", alpha=0.6)
    path = run_gd(lr)
    ax.plot(path[:, 0], path[:, 1], "o-", color="crimson", ms=4)
    ax.plot(*path[0], "g*", ms=15, label="start")
    ax.plot(0, 0, "b*", ms=15, label="minimum")
    ax.set_title(f"learning rate = {lr}")
    ax.set_xlabel("w1")
axes[0].set_ylabel("w2")
axes[0].legend()
fig.suptitle("Effect of the Learning Rate (too small: slow, too large: oscillation)", fontsize=13)
fig.tight_layout()
plt.show()
````

### Project 7: PCA projection of high-dimensional data
Reduce 10D data to 2D with PCA (via SVD) and visualize the clusters.

````python
rng = np.random.default_rng(14)
centers = rng.normal(0, 4, size=(3, 10))
data = np.vstack([rng.normal(c, 1.0, size=(100, 10)) for c in centers])
labels = np.repeat([0, 1, 2], 100)

# PCA by SVD
Xc = data - data.mean(axis=0)
U, S, Vt = np.linalg.svd(Xc, full_matrices=False)
proj = Xc @ Vt[:2].T
explained = (S ** 2) / (S ** 2).sum()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.2))

sc = ax1.scatter(proj[:, 0], proj[:, 1], c=labels, cmap="Set1", edgecolor="k", linewidth=0.3, alpha=0.8)
ax1.set_xlabel(f"PC1 ({explained[0]:.0%})")
ax1.set_ylabel(f"PC2 ({explained[1]:.0%})")
ax1.set_title("PCA Projection (10D -> 2D)")

ax2.bar(range(1, 11), explained, color="steelblue", label="Individual")
ax2.plot(range(1, 11), np.cumsum(explained), "ro-", label="Cumulative")
ax2.set_xlabel("Principal component")
ax2.set_ylabel("Explained variance ratio")
ax2.set_title("Explained Variance")
ax2.legend()

fig.tight_layout()
plt.show()
````

### Project 8: Grid of sample images with labels
Typical for inspecting image datasets. We create tiny synthetic "digit-like" images.

````python
rng = np.random.default_rng(15)

def make_image(kind):
    img = np.zeros((16, 16))
    if kind == 0:      # square
        img[4:12, 4:12] = 1
    elif kind == 1:    # horizontal bar
        img[7:9, 2:14] = 1
    elif kind == 2:    # vertical bar
        img[2:14, 7:9] = 1
    else:              # diagonal
        np.fill_diagonal(img, 1)
    return np.clip(img + rng.normal(0, 0.15, img.shape), 0, 1)

names = ["square", "h-bar", "v-bar", "diagonal"]
kinds = rng.integers(0, 4, size=12)

fig, axes = plt.subplots(3, 4, figsize=(8, 6.2))
for ax, k in zip(axes.flat, kinds):
    ax.imshow(make_image(k), cmap="gray")
    ax.set_title(f"label: {names[k]}", fontsize=10)
    ax.axis("off")
fig.suptitle("Sample Images from a Dataset", fontsize=13)
fig.tight_layout()
plt.show()
````

### Project 10: Animation-style view of model fitting (polynomial underfit vs. overfit)
Fit polynomials of different degrees and compare. This shows the **bias-variance trade-off** visually.

````python
rng = np.random.default_rng(17)
x_train = np.sort(rng.uniform(0, 1, 15))
y_train = np.sin(2 * np.pi * x_train) + rng.normal(0, 0.25, 15)
x_plot = np.linspace(0, 1, 300)

degrees = [1, 4, 14]
titles = ["Underfit (degree 1)", "Good fit (degree 4)", "Overfit (degree 14)"]

fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)
for ax, d, t in zip(axes, degrees, titles):
    coeffs = np.polyfit(x_train, y_train, d)
    ax.scatter(x_train, y_train, color="k", label="Training data", zorder=3)
    ax.plot(x_plot, np.sin(2 * np.pi * x_plot), "g--", label="True function")
    ax.plot(x_plot, np.polyval(coeffs, x_plot), "r", lw=2, label="Model")
    ax.set_ylim(-1.8, 1.8)
    ax.set_title(t)
axes[0].legend()
fig.tight_layout()
plt.show()
````

### Best practices
- Prefer the **object-oriented interface** (`fig, ax = plt.subplots()`).
- Always label **axes, units, and titles**, and add a **legend** when there are several series.
- Pick the **right chart**: line (trends), scatter (relationships), bar (categories), histogram (distributions), heatmap (matrices).
- Choose **perceptually uniform colormaps** (`viridis`) and use diverging maps (`coolwarm`) for data centered on zero.
- Use **log scales** for values spanning orders of magnitude.
- Use **vector formats** (`.pdf`, `.svg`) for papers and `dpi=300` for raster images.
- Do not hide uncertainty: show **error bars or confidence bands** when you average over runs.
